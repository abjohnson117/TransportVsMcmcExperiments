```python
from typing import Callable, List

class MLP(eqx.Module):
    layers: List[eqx.nn.Linear]         # main hidden layers
    norms:  List[eqx.nn.LayerNorm]      # per-layer norms (stateless, no BN state)
    skips:  List[eqx.nn.Linear | None]  # projections for residuals (or None for identity)
    out:    eqx.nn.Linear
    activation_fn: List[Callable]

    def __init__(
        self,
        key: jax.random.PRNGKey,
        dim: int,
        out_dim: int | None = None,
        num_layers: int = 4,
        activation_fn: List[Callable] | Callable = jax.nn.gelu,
        w: int | List[int] = 64,
        time_varying: bool = False,
    ):
        if out_dim is None:
            out_dim = dim

        # normalize activation list
        if isinstance(activation_fn, list):
            if len(activation_fn) == 1:
                activation_fn *= num_layers - 1
        else:
            activation_fn = [activation_fn] * (num_layers - 1)
        self.activation_fn = activation_fn

        # normalize widths
        if isinstance(w, list):
            if len(w) == 1:
                w *= (num_layers - 1)
            widths = w
        else:
            widths = [w] * (num_layers - 1)

        k = jax.random.split(key, 2 * num_layers + 1)

        in_dim0 = dim + (1 if time_varying else 0)

        self.layers = []
        self.norms  = []
        self.skips  = []

        in_dim = in_dim0
        for i, width in enumerate(widths):
            # main linear
            lin = eqx.nn.Linear(in_dim, width, key=k[i])
            self.layers.append(lin)

            # lightweight, stateless normalization over features
            self.norms.append(eqx.nn.LayerNorm(width))

            # residual projection if needed
            if in_dim == width:
                self.skips.append(None)
            else:
                self.skips.append(eqx.nn.Linear(in_dim, width, key=k[i + num_layers]))

            in_dim = width

        # output layer
        self.out = eqx.nn.Linear(in_dim, out_dim, key=k[-1])

    def __call__(self, x):
        for layer, norm, skip, act in zip(self.layers, self.norms, self.skips, self.activation_fn):
            h = layer(x)
            h = norm(h)      # normalize pre-activation
            h = act(h)
            s = x if skip is None else skip(x)
            x = h + s
        return self.out(x)
```

```python
def read_data_h5(path="data.h5"):
    with h5py.File(path, "r") as f:
        targets = f["/target"][...]
        data = f["/data"][...]
    return targets, data
```

```python
with open("poisson.yaml") as fid:
    inargs = yaml.full_load(fid)

map_est = np.load("training_dataset/map_param_grid.npy")
targets, yobs = read_data_h5()
```

```python
nsamples = 10000
nx = ny = 33

def load_hmala_chains(chain_dir, n_chains=5, thin=50):
    chains = []
    for i in range(n_chains):
        path = os.path.join(chain_dir, f"chain_{i:03d}", "hmala_samples.npz")
        arr = np.load(path)['arr']  # (100000, 1089)
        chains.append(arr[::thin])  # 2000 samples per chain
    return np.vstack(chains)  # (10000, 1089) total

hmala_samps_flat = load_hmala_chains("mcmc_main_converge_samps")
hmala_samps = hmala_samps_flat.reshape(nsamples, nx, ny)
hmala_mean = np.mean(hmala_samps, axis=0)
hmala_var  = np.var(hmala_samps, axis=0)
```

```python
train_dim = 1024 * 4
nx = ny = 33
flat_length = nx * ny
ys = (np.load("training_dataset/solutions_grid_delta.npy"))[:train_dim]
us = (np.load("training_dataset/parameters_delta.npy"))[:train_dim, :].reshape(train_dim, flat_length)

us_ref = us.copy()
np.random.shuffle(us_ref)

us_test = (np.load("training_dataset/parameters_delta.npy"))[train_dim:train_dim + 20000, :].reshape(20000, flat_length)
np.random.shuffle(us_test)

ys_normalizer = UnitGaussianNormalizer(ys)
ys_normalized = ys_normalizer.encode()
```

```python
mean_us = us.mean(axis=0)
std_us  = us.std(axis=0) + 1e-8   # per-feature standardisation (no PCA)

def param_encode(b):
    return (b - mean_us) / std_us

def param_decode(z):
    return z * std_us + mean_us
```

```python
us_enc     = param_encode(us)
us_ref_enc = param_encode(us_ref)
us_test_enc = param_encode(us_test)

target_data = jnp.hstack([jnp.asarray(ys_normalized), jnp.asarray(us_enc)])
ref_data    = jnp.hstack([jnp.asarray(ys_normalized), jnp.asarray(us_ref_enc)])
```

```python
# --- Model definition (no PCA: u-dim = flat_length = 1089) ---
key = random.PRNGKey(seed=42)
key1, key2 = random.split(key=key, num=2)

batch_size  = 2048
steps       = 5000
print_every = 10000

yu_dimension = (yobs.shape[0], flat_length)   # (100 obs, 1089 params)
dim          = yu_dimension[0] + yu_dimension[1]

hidden_layer_list = [256] * 3
model = MLP(
    key=key2,
    dim=dim,
    time_varying=True,
    w=hidden_layer_list,
    num_layers=len(hidden_layer_list) + 1,
    activation_fn=jax.nn.gelu,
)

schedule = optax.warmup_cosine_decay_schedule(
    init_value=0.0,
    peak_value=3e-4,
    warmup_steps=2_000,
    decay_steps=steps,
    end_value=1e-5,
)
optimizer       = optax.chain(optax.clip_by_global_norm(1.0), optax.adamw(schedule))
interpolant     = trig_interpolant
interpolant_der = trig_interpolant_der
interpolant_args = {"t": None, "x1": None, "x0": None}
```

```python
# --- Train ---
trainer = NNTrainer(
    target_density=None,
    model=model,
    optimizer=optimizer,
    interpolant=interpolant,
    interpolant_der=interpolant_der,
    reference_sampler=gaussian_reference_sampler,
    loss=vec_field_loss,
    interpolant_args=interpolant_args,
    yu_dimension=yu_dimension,
)
trainer.train(
    train_data=target_data,
    train_dim=train_dim,
    batch_size=batch_size,
    steps=steps,
    x0_data=ref_data,
    print_every=print_every,
);
```

```python
# --- Generate posterior samples conditioned on yobs ---
yobs            = np.load("data_obs.npy")
yobs_normalized = ys_normalizer.encode(yobs)

cond_samples = trainer.conditional_sample(
    cond_values=yobs_normalized, u0_cond=jnp.asarray(us_test_enc), nsamples=len(us_test_enc)
)
u_flat    = param_decode(np.array(cond_samples[:, yu_dimension[0]:]))

u_samples = jnp.asarray(u_flat.reshape(-1, nx, ny))
u_means   = jnp.mean(u_samples, axis=0)
u_var     = jnp.var(u_samples, axis=0)
```

```python
# --- Posterior covariance eigenvalue spectra ---
plt.style.use("ggplot")
si_flat    = np.array(u_samples).reshape(u_samples.shape[0], -1)   # transport
hmala_flat = hmala_samps_flat                                       # h-MALA reference

si_c    = si_flat    - si_flat.mean(axis=0)
hmala_c = hmala_flat - hmala_flat.mean(axis=0)

C_si    = (si_c.T    @ si_c)    / (si_c.shape[0]    - 1)
C_hmala = (hmala_c.T @ hmala_c) / (hmala_c.shape[0] - 1)

eigvals_si    = np.linalg.eigvalsh(C_si)[::-1]
eigvals_hmala = np.linalg.eigvalsh(C_hmala)[::-1]

n_plot = 30
k_line = 20
idx    = np.arange(1, n_plot + 1)

fig, ax = plt.subplots(figsize=(7, 4))
ax.semilogy(idx, eigvals_hmala[:n_plot], color="tab:red",  lw=2.5, label="h-MALA")
ax.semilogy(idx, eigvals_si[:n_plot],    color="tab:blue", lw=2.5, label="Transport", linestyle="--")
# ax.axvline(k_line, color="gray", linestyle="--", lw=1.5, label=f"k = {k_line}")
ax.set_xlabel("Mode index", fontsize=13)
ax.set_ylabel("Eigenvalue (log scale)", fontsize=13)
ax.set_title("Eigenvalue spectra", fontsize=14)
ax.legend(fontsize=11)
ax.set_xlim(1, n_plot)
ax.grid(True, which="both", alpha=0.3)
plt.tight_layout()
# plt.savefig("images/eigenvalue_decay.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Load SI-ODE results from most recent sample_convergence run (convergence_results_final)
input_dir = "convergence_results_final"
sample_no_list_conv = np.load(os.path.join(input_dir, "run_00", "sample_no_list.npy"))
train_idx = int(np.searchsorted(sample_no_list_conv, 50000))  # index for 50000-sample entry
ntrials = 10

def load_si_mean_var(scenario):
    means = np.stack([
        np.load(os.path.join(input_dir, f"run_{i:02d}", f"u_mean_{scenario}.npy"))[train_idx]
        for i in range(ntrials)
    ])
    vars_ = np.stack([
        np.load(os.path.join(input_dir, f"run_{i:02d}", f"u_var_{scenario}.npy"))[train_idx]
        for i in range(ntrials)
    ])
    return np.mean(means, axis=0), np.mean(vars_, axis=0)

si_mean_obs, si_var_obs = load_si_mean_var("obs")
si_mean_98,  si_var_98  = load_si_mean_var("98")
si_mean_med, si_var_med = load_si_mean_var("med")
```

```python
k = get_gaussianRBF(10.0)

k = vectorize_kfunc(k)

@jax.jit
def MMD(X, Y):
    # print(X.shape, Y.shape)
    return jnp.mean(k(X, X)) + jnp.mean(k(Y, Y)) - 2 * jnp.mean(k(X, Y))

MMD(u_samples, hmala_samps)
```
