```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("ggplot")
```

```python
us_all = (np.load("training_dataset/parameters.npy")
          .reshape(-1, 21, 21, 3, order="F")[:, :, :, 0]
          .reshape(-1, 21 * 21))   
us_delt = (np.load("training_dataset/parameters_delta.npy")
          .reshape(-1, 21, 21, 3, order="F")[:, :, :, 0]
          .reshape(-1, 21 * 21))  
```

```python
idx_no = 199
u = us_all[idx_no]
u_delt = us_delt[idx_no]
u_norm = np.linalg.norm(u)
u_norm_delt = np.linalg.norm(u_delt)
print(f"This is the norm of u: {u_norm.item()}")
print(f"This is the norm of u_delta: {u_norm_delt.item()}")
```

```python
us_std = np.std(us_all, axis=0)
us_delt_std = np.std(us_delt, axis=0)
plt.scatter(us_std, us_delt_std)
plt.plot(np.linspace(np.min(us_std), np.max(us_std), 800), np.linspace(np.min(us_std), np.max(us_std), 800), c="black", label=r"$y = x$")
plt.xlabel(r"$\delta = 1$ std")
plt.ylabel(r"$\delta = 9$ std")
plt.legend()
```
